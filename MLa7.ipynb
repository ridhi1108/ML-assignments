```python

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

# Step 1: Load the dataset
data = pd.read_csv("purchase_logistic.csv")

# Display dataset
print("Dataset:")
print(data.head())

# Step 2: Select independent and dependent variables
X = data[['Age', 'EstimatedSalary']]
y = data['Purchased']

# Step 3: Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Step 4: Feature scaling
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Step 5: Create Gaussian Naive Bayes classifier
classifier = GaussianNB()

# Step 6: Train the model
classifier.fit(X_train, y_train)

# Step 7: Predict the results
y_pred = classifier.predict(X_test)

# Step 8: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)

# Step 9: Display confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

# Step 10: Plot confusion matrix
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Not Purchased', 'Purchased'],
    yticklabels=['Not Purchased', 'Purchased']
)

plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')
plt.title('Confusion Matrix - Question 1')
plt.show()

# Step 11: Print classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

# Step 1: Load the dataset
data = pd.read_csv("purchase_logistic.csv")

# Display dataset
print("Dataset:")
print(data.head())

# Step 2: Encode Gender
le = LabelEncoder()
data['Gender'] = le.fit_transform(data['Gender'])

print("\nGender Encoding:")
print(dict(zip(le.classes_, le.transform(le.classes_))))

# Step 3: Select independent and dependent variables
X = data[['Gender', 'Age', 'EstimatedSalary']]
y = data['Purchased']

# Step 4: Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Step 5: Feature scaling
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Step 6: Create Gaussian Naive Bayes classifier
classifier = GaussianNB()

# Step 7: Train the model
classifier.fit(X_train, y_train)

# Step 8: Predict the results
y_pred = classifier.predict(X_test)

# Step 9: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)

# Step 10: Display confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

# Step 11: Plot confusion matrix
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Greens',
    xticklabels=['Not Purchased', 'Purchased'],
    yticklabels=['Not Purchased', 'Purchased']
)

plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')
plt.title('Confusion Matrix - Question 2')
plt.show()

# Step 12: Print classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
