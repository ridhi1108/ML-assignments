```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Load dataset
df = pd.read_csv("USA_Housing.csv")

# Features and target
X = df.drop(["Address", "Price"], axis=1)
y = df["Price"]

best_r2 = -1
best_state = 0

# Try 200 random states
for state in range(200):

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, random_state=state
    )

    # Normalization
    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    # Model
    model = LinearRegression()
    model.fit(X_train, y_train)

    pred = model.predict(X_test)

    r2 = r2_score(y_test, pred)

    if r2 > best_r2:
        best_r2 = r2
        best_state = state
        best_model = model
        best_pred = pred
        best_ytest = y_test

# Results
print("Best Random State =", best_state)
print("Best R2 Score =", best_r2)

print("\nIntercept (β0):", best_model.intercept_)

print("\nCoefficients")
for name, coef in zip(X.columns, best_model.coef_):
    print(name, "=", coef)

mae = mean_absolute_error(best_ytest, best_pred)
rmse = np.sqrt(mean_squared_error(best_ytest, best_pred))

print("\nMAE =", mae)
print("RMSE =", rmse)
print("R2 =", best_r2)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

df = pd.read_csv("USA_Housing.csv")

X = df[[
    "Avg. Area Income",
    "Avg. Area House Age",
    "Area Population"
]]

y = df["Price"]

best_r2 = -1

for state in range(200):

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, random_state=state
    )

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    model = LinearRegression()

    model.fit(X_train, y_train)

    pred = model.predict(X_test)

    r2 = r2_score(y_test, pred)

    if r2 > best_r2:
        best_r2 = r2
        best_state = state
        best_model = model
        best_pred = pred
        best_ytest = y_test

print("Best Random State =", best_state)

print("\nIntercept =", best_model.intercept_)

print("\nCoefficients")

for name, coef in zip(X.columns, best_model.coef_):
    print(name, "=", coef)

mae = mean_absolute_error(best_ytest, best_pred)
rmse = np.sqrt(mean_squared_error(best_ytest, best_pred))

print("\nMAE =", mae)
print("RMSE =", rmse)
print("R2 =", best_r2)
```
