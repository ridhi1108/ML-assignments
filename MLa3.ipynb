```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset("titanic")
print(titanic.head())
```

```python
sns.scatterplot(data= titanic, x="age", y="fare")
plt.title("Age vs Fare")
plt.show()
```

```python
sns.scatterplot(
    data= titanic,
    x= "age",
    y= "fare",
    hue="survived"
)
plt.show()
```

```python
sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    style="sex",
    hue="sex"
)

plt.show()
```

```python
sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    size="pclass",
    hue="pclass"
)

plt.show()
```

```python
sns.lineplot(
    data=titanic,
    x="pclass",
    y="fare"
)

plt.title("Passenger Class vs Fare")
plt.show()
```

```python
sns.relplot(
    data=titanic,
    x="age",
    y="fare",
    hue="survived"
)

plt.show()
```

```python
sns.relplot(
    data=titanic,
    x="pclass",
    y="fare",
    kind="line"
)

plt.show()
```

```python
sns.pairplot(
    titanic,
    hue="survived"
)

plt.show()
```

```python
g = sns.FacetGrid(
    titanic,
    col="sex"
)
g.map_dataframe(
    sns.scatterplot,
    x="age",
    y="fare"
)
plt.show()
```

```python
g = sns.PairGrid(titanic[["age", "fare", "sibsp", "parch"]])

g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.map_diag(sns.histplot)

plt.show()
```

```python
g = sns.JointGrid(
    data= titanic,
    x="age",
    y="fare"
)
g.plot(
    sns.scatterplot,
    sns.histplot
)
plt.show()
```

```python
flights = sns.load_dataset("flights")
print(flights.head())
```

```python
sns.lineplot(
    data = flights,
    x="year",
    y="passengers"
)
plt.title("Passengers over years")
plt.show()
```

```python
sns.lineplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month"
)

plt.show()
```

```python
sns.scatterplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month"
)

plt.show()
```

```python
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month"
)

plt.show()
```

```python
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    kind="line",
    hue="month"
)

plt.show()
```

```python
g = sns.FacetGrid(
    flights,
    col="month",
    col_wrap=4
)

g.map_dataframe(
    sns.lineplot,
    x="year",
    y="passengers"
)

plt.show()
```
