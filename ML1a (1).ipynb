```python
# Create a 10×10 random array.
# Use seed() so the values remain the same.
# Slice three arrays:
# A → 5×7
# B → 7×5
# C → 5×7
# Find A + C
# Find A × B 
import numpy as np
np.random.seed(10)
arr= np.random.randint(1,100, size=(10,10))
print("Original 10 x 10 Array:")
print(arr)

A = arr[0:5, 0:7]
B = arr[0:7, 0:5]
C = arr[5:10, 0:7]

print("\nArray A (5 x 7):")
print(A)

print("\nArray B (7 x 5):")
print(B)

print("\nArray C (5 x 7):")
print(C)

# Addition of A and C
sum_AC = A + C
print("\nSum of A and C:")
print(sum_AC)

# Matrix multiplication of A and B
multiply_AB = A @ B
print("\nMultiplication of A and B:")
print(multiply_AB) 
```

```python
import numpy as np
np.random.seed(20)
arr = np.random.randint(1, 100, size=(10, 10))
print(arr)
#to print total elements
n = arr.size
print("\nTotal elements:", n)

shapes_2d = []
for i in range(1, n + 1):
    if n % i == 0:
        j = n // i
        shapes_2d.append((i, j))
print("\nAll possible 2-D shapes:")
print(shapes_2d)

shapes_3d = []
for i in range(1, n + 1):
    for j in range(i, n + 1):
        if n % (i * j) == 0:
            k = n // (i * j)

            if j <= k:
                shapes_3d.append((i, j, k))
print("\nAll possible 3-D shapes:")
print(shapes_3d)
```
